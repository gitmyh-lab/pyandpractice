```python
import threading
import time

#线程体函数
def thread_body():
    t = threading.current_thread()
    for n in range(5):
        print('第{0}次执行线程{1}'.format(n, t.name))
        time.sleep(1)
    print('线程{0}执行完成！'.format(t.name))

def main():
    t1 = threading.Thread(target=thread_body)
    t1.start()

    t2 = threading.Thread(target=thread_body, name='MyThread')
    t2.start()

if __name__ == '__main__':
    main()
```

```python
import threading
import time
import random

#创建条件变量对象
condition = threading.Condition()

class Stack():
    def __init__(self):
        #堆栈指针初始为零
        self.pointer = 0
        #堆栈由5个数字空间，-1表示没有数据
        self.data = [-1, -1, -1, -1, -1]

    def push(self, c):
        global condition
        condition.acquire() #代码需要同步
        #堆栈已满，不能压栈
        while self.pointer == len(self.data):
            condition.wait() #等待其他线程把数据出栈
        condition.notify() #通知其他线程把数据出栈
        self.data[self.pointer] = c #数据压栈
        self.pointer += 1 #指针向上移动
        condition.release()

    def pop(self):
        global condition
        condition.acquire()
        #堆栈无数据，不能出栈
        while self.pointer == 0:
            #等待其他线程把数据压栈
            condition.wait()
        condition.notify() #通知其他线程压栈
        self.pointer -= 1
        data = self.data[self.pointer]
        condition.release()
        return data #数据出栈

#创建堆栈对象
stack = Stack()

#生产者线程体函数
def producer_thread_body():
    global stack
    for i in range(0, 10):
        stack.push(i)
        print('生产：{0}'.format(i))
        time.sleep(1)

def consumer_thread_body():
    global stack
    for i in range(0, 10):
        x = stack.pop()
        print('消费：{0}'.format(x))
        time.sleep(1)

def main():
    producer = threading.Thread(target=producer_thread_body)
    producer.start()
    consumer = threading.Thread(target=consumer_thread_body)
    consumer.start()

if __name__ == '__main__':
    main()
```

```python
import threading
import time
import random

event = threading.Event()

class Stack():
    def __init__(self):
        #堆栈指针初始为零
        self.pointer = 0
        #堆栈由5个数字空间，-1表示没有数据
        self.data = [-1, -1, -1, -1, -1]

    def push(self, c):
        global event
        #堆栈已满，不能压栈
        while self.pointer == len(self.data):
            event.wait() #等待其他线程把数据出栈
        event.set() #通知其他线程把数据出栈
        self.data[self.pointer] = c #数据压栈
        self.pointer += 1 #指针向上移动

    def pop(self):
        global event
        #堆栈无数据，不能出栈
        while self.pointer == 0:
            #等待其他线程把数据压栈
            event.wait()
        event.set() #通知其他线程压栈
        self.pointer -= 1
        data = self.data[self.pointer]
        return data #数据出栈

#创建堆栈对象
stack = Stack()

#生产者线程体函数
def producer_thread_body():
    global stack
    for i in range(0, 10):
        stack.push(i)
        print('生产：{0}'.format(i))
        time.sleep(1)

def consumer_thread_body():
    global stack
    for i in range(0, 10):
        x = stack.pop()
        print('消费：{0}'.format(x))
        time.sleep(1)

def main():
    producer = threading.Thread(target=producer_thread_body)
    producer.start()
    consumer = threading.Thread(target=consumer_thread_body)
    consumer.start()

if __name__ == '__main__':
    main()
```
