```python
# 有条件查询
import pymysql

connection = pymysql.connect(host='localhost',
                            user='root',
                            password='123456',
                            database='MyDB',
                            charset='utf8')

try:
    with connection.cursor() as cursor:
        sql = 'select name, userid from user where userid >%(id)s'
        cursor.execute(sql, {'id':0})
        result_set = cursor.fetchall()
        for row in result_set: #字段的顺序是select语句中列出的字段顺序
            print('id:{0}-name{1}'.format(row[1],row[0]))
finally:
    connection.close()
```

```python
#无条件查询
import pymysql

connection = pymysql.connect(host='localhost',
                            user='root',
                            password='123456',
                            database='MyDB',
                            charset='utf8')
try:
    with connection.cursor() as cursor:
        sql = 'select max(userid) from user'
        cursor.execute(sql)
        row = cursor.fetchall()
        if row is not None:
            print('最大用户id：{0}'.format(row[0]))
finally:
    connection.close()
    
```
