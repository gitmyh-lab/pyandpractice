```python
#同一性测试运算符有两个： not
class Person : 
    def __init__(self, name, age):
        self.name = name 
        self . age= age 
    def __eq__(self, other): 
        if self.name ==other.name and self.age == other.age: 
            return True 
        else:
            return False 
pl = Person('Tony', 18) 
p2 = Person('Tony', 18) 
print(pl == p2) 
print(pl is p2) 
print(pl != p2) 
print(pl is not p2) 
```

```python
for item in range(10): 
    if item == 3: 
        #跳出循环
        break 
    print ('Count is : {0}'.format(item))
```

```python
for item in range(10): 
    if item == 3: 
        #结束本次循环
        continue 
    print ('Count is : {0}'.format(item))
```
